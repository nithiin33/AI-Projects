```python
import os
from dotenv import load_dotenv
from crewai import Agent, Task, Crew, Process, LLM

# Load environment variables
load_dotenv()

# 1. Configure the LLM (Foundation Layer)
groq_llm = LLM(
    model="groq/llama-3.1-8b-instant",
    api_key=os.environ.get("GROQ_API_KEY")
)

# 2. Create Specialized Agents (Behavior Layer)
res = Agent(
    role="Research Analyst",
    goal="Conduct structured research on a topic",
    backstory="As a Research Analyst, I meticulously gather and analyze information to provide accurate insights.",
    llm=groq_llm
)

wri = Agent(
    role="Technical Writer",
    goal="Write a detailed and clear blog post",
    backstory="As a Technical Writer, I transform complex ideas into clear and engaging content for a broad audience.",
    llm=groq_llm
)

edi = Agent(
    role="Content Editor",
    goal="Refine and polish the content professionally",
    backstory="As a Content Editor, I ensure all written material is polished, coherent, and meets high editorial standards.",
    llm=groq_llm
)

# 3. Define Tasks (Execution Layer)
t1 = Task(
    description="Research how Multi-Agent AI systems work.",
    agent=res,
    expected_output="Structured research notes."
)

t2 = Task(
    description="Write a blog post based on research.",
    agent=wri,
    expected_output="Well-structured blog post."
)

t3 = Task(
    description="Edit and polish the blog post.",
    agent=edi,
    expected_output="Final polished blog post."
)

# 4. Orchestrate with a Crew (Sequential Process)
crew = Crew(
    agents=[res, wri, edi],
    tasks=[t1, t2, t3],
    process=Process.sequential
)

# 5. Execute the System
print("Starting the multi-agent process...")
result = crew.kickoff()
print("\nFinal Polished Output:\n")
print(result)
```
